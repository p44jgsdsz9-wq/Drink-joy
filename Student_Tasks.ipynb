```python

tasks = []

def add_task():
    task = input("Enter a task: ")
    tasks.append(task)
    print("Task added successfully!")

def show_tasks():
    print("\nYour Tasks:")
    for task in tasks:
        print("-", task)

add_task()
add_task()
show_tasks()
```
